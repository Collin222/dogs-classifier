```python
import torch
import torchvision.models as models
import json
```

```python
obj = torch.load("vgg11.pth", map_location="cpu")
keys = list(dict.keys(obj))
print(keys)
```

```python
output = {}

for key in keys:
    t = obj[key]
    print(key, t.shape)
    output[key] = t.tolist()
```

```python
with open("vgg11weights.json", "w") as f:
    json.dump(output, f)
```
